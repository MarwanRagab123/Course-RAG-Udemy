## First Embedding Vectors

```python
# We use Hugging Face And OpenAi
from sentence_transformers import SentenceTransformer

embeddings = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

embeddings
```

```python
#Create Your First Embedding
text = "This is a test document"
embedding = embeddings.encode(text)
print(f"Length of embedding: {len(embedding)}")
print(f"Embedding: {embedding}")
```

```python
sentences = [
    "This is a test document",
    "This is another test document",
    "ArithmeticError is a base class for errors that occur for numeric calculations.",
]
sentence_embeddings = embeddings.encode(sentences)
print(f"Length of sentence embedding: {len(sentence_embeddings)}")
print(f"Sentence Embeddings[1]: {len(sentence_embeddings[1])}")
#summary
'''
 We have created a sentence transformer model using Hugging Face and OpenAI.
 We have created an embedding for a single text and for a list of sentences.
 The embedding is a vector representation of the text that can be used for various applications such as similarity search, clustering, and classification.
 each chunks of text is converted into a vector of numbers that captures the semantic meaning of the text.
'''
```
