## Handling PDF Challenges
*-store text in complex ways(not just simple text)*

*-Can have formmating issues*

*-May contined Scanned Image(require OCR)*

*-often have extractions artifacts*

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
class RAGPDFProcess:
    '''Advanced PDF processing'''
    def __init__(self, chunk_size = 1000, chunk_overlab = 200):
        self.chunk_size = chunk_size
        self.chunk_overlab = chunk_overlab
        self.text_splitter = RecursiveCharacterTextSplitter(chunk_size = chunk_size,
         chunk_overlap = chunk_overlab,
         separators = ["\n\n", "\n", " ", ""]
         )
    
    def Process_Pdf(self, path:str):

        #Load Pdf
        try:
            pypdf = PyPDFLoader(path)
            pages = pypdf.load()
        except Exception as e:
            print(f"Erorr is {e}")
        #process each page

        processed_chunks = []

        for num, page in enumerate(pages):
            #clean text
            cleaned_text = self._clean_text(page.page_content)

            #check empty images

            if len(cleaned_text.strip())<50:
                continue

            #create chunks and enhance meta data

            chunks = self.text_splitter.create_documents(
                texts=[cleaned_text],
                metadatas=[{
                    **page.metadata,
                    "page_num":num + 1,
                    "total_page":len(pages),
                    "chunk_method":"smarter pdf process",
                    "char_count":len(cleaned_text)
                }]
            )
            processed_chunks.extend(chunks)

        return processed_chunks
        


    def _clean_text(self, text):
        text = " ".join(text.split())
        return text
    
```

```python
prerprocess = RAGPDFProcess()
```

```python
chunks = prerprocess.Process_Pdf("data/Final_FTAI_SpaVagile_conference-template-a4_6pages.pdf")
```

```python
for i ,val in enumerate(chunks[0].metadata.items()):
    print(f"{i}:{val}")
```

```python
for i ,val in enumerate(chunks):
    print(f"Size of this chunk = {len(val.page_content)} characters")
    print(f"Chunk #{i+1} Content: {val.page_content}")
```
