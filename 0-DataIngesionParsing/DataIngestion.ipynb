```python
sample_text = {
    "data/text_files/python_intro.txt":''' Python is a high-level programming language known for being simple and easy to read. It’s widely used in:

AI & Machine Learning (your field 👀)
Web development (e.g., Flask, Django)
Data analysis
Automation & scripting''',

"data/text_files/machine_learning.txt":''' Machine Learning (ML) is a part of AI where systems learn patterns from data instead of being explicitly programmed.

In short:
You give the model data
It learns patterns
It makes predictions or decisions
Main types:
Supervised → learns from labeled data (e.g., fraud detection)
Unsupervised → finds patterns without labels (e.g., clustering)
Reinforcement → learns by trial and error (rewards & penalties)'''
}

for path, content in sample_text.items():
    with open(path,"w", encoding="UTF-8") as f:
        f.write(content)

print("Sample Text Created!")
```

#### Text Loader - single file

```python
from langchain_community.document_loaders import TextLoader
loader = TextLoader("data/text_files/machine_learning.txt",encoding = "utf-8")
docs2 = loader.load()

print(f"length of document = {len(docs2)}")
print(f"content is {docs2[0].page_content[:100]}")
print(f"metadata {docs2[0].metadata}")
```

### TextLoader - Multiple file

```python
from langchain_community.document_loaders import DirectoryLoader

ldir_oader = DirectoryLoader(
    "data/text_files",
    glob = "**/*.txt",
    loader_cls=TextLoader,
    loader_kwargs={'encoding':'utf-8'},
)

docs = ldir_oader.load()

for i , doc in enumerate(docs):
    print(f"content document {i+1} :")
    print(f"content is {doc.metadata['source']}")
    print(f"Length = {len(doc.page_content)}")
```

### Split Documents into chunks

```python
from langchain_core.documents import Document
from langchain_text_splitters import (
    RecursiveCharacterTextSplitter,
    CharacterTextSplitter,
    TokenTextSplitter,
)
```

```python

char_splitter = CharacterTextSplitter(
    separator = "\n",
    chunk_size = 200,
    chunk_overlap =20
)

chunks = char_splitter.split_documents(docs)

chunks
for i, con in enumerate(chunks):
    print(f"chunks in document {i+1} are {con.page_content[0:50]}")
```

```python
rec_splitter = RecursiveCharacterTextSplitter(
    separators = ["\n\n"," ","،","\n"],
    chunk_size = 200,
    chunk_overlap = 30
)

chu = rec_splitter.split_documents(docs2)

for i , cont in enumerate(chu):
    print(f"chunk content {i+1} is {cont.page_content}")
    print()
```
